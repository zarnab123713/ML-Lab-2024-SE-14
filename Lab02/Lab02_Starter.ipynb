## Task 6 · Census 2023 (real data, PBS)
Using `units`, `pop_2023` and `pop_2017`:

- **(c)** The most populous unit.

- **(f)** The fastest-growing unit.

**Interpret:** in 1–2 lines, what do (e) and (f) say? Compare AJK with the rest.

```python
import numpy as np

# Given datasets
units = np.array([
    "Punjab", "Sindh", "Khyber Pakhtunkhwa",
    "Balochistan", "Islamabad", "AJK", "Gilgit-Baltistan"
])

pop_2023 = np.array([
    127_688_922, 55_696_147, 40_856_097,
    14_894_402, 2_363_863, 4_333_467, 1_709_049
])

pop_2017 = np.array([
    109_989_655, 47_854_510, 35_501_964,
    12_335_129, 2_003_368, 4_045_367, 1_492_924
])
# Task6a: Total population.
# Calculate total population
total_population = np.sum(pop_2023)

print("Task6a: Total Population:")
print(total_population)
# Task6b: Each unit's share of the total, in %.
# Calculate each unit's share in percentage
shares = (pop_2023 / total_population) * 100

print("\nTask6b: Population Share:")
for unit, share in zip(units, shares):
    print(unit, ":", round(share, 2), "%")
# Task6c: The most populous unit.
# Find the most populous unit
most_populous = np.argmax(pop_2023)

print("\nTask6c: Most Populous Unit:")
print(units[most_populous])
# Task6d: Units with more than 20 million people.
# Find units with more than 20 million people
above_20_million = units[pop_2023 > 20_000_000]

print("\nTask6d: Units with More Than 20 Million:")
print(above_20_million)
# Task6e: Annual growth rate 2017–2023 for every unit, in one line: ((pop_2023 / pop_2017) ** (1/6) - 1) * 100.
#  Calculate annual growth rate (2017–2023)
growth_rate = ((pop_2023 / pop_2017) ** (1/6) - 1) * 100

print("\nTask6e: Annual Growth Rate:")
for unit, rate in zip(units, growth_rate):
    print(unit, ":", round(rate, 2), "%")
# Task6f:The fastest-growing unit.
#  Find the fastest-growing unit
fastest = np.argmax(growth_rate)

print("\nTask6f: Fastest-Growing Unit:")
print(units[fastest])
print("Growth Rate:", round(growth_rate[fastest], 2), "%")
```

##### Interpret: in 1–2 lines, what do (e) and (f) say? Compare AJK with the rest.

- Part (e) shows the annual population growth rate of each region from 2017 to 2023.
- Part (f) shows that Balochistan has the highest annual growth rate (3.19%), while AJK has the lowest (1.15%).

AJK Comparison: AJK's population is growing more slowly than all the other regions.
Its annual growth rate (1.15%) is much lower than the average growth rate of the other six regions (2.62%).

```python
# D3. CPI inflation, year-on-year % (Pakistan Economic Survey 2024-25, Fig 7.1, from PBS)
cpi_months = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec", "Jan", "Feb", "Mar", "Apr"])
cpi = np.array([[28.3, 27.4, 31.4, 26.8, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3],    # FY2024
                [11.1,  9.6,  6.9,  7.2,  4.9,  4.1,  2.4,  1.5,  0.7,  0.3]])   # FY2025
#Task8a:Average inflation of each year.
average_inflation = np.mean(cpi, axis=1)

print("Task8a:")
print("FY2024 Average Inflation:", round(average_inflation[0], 2), "%")
print("FY2025 Average Inflation:", round(average_inflation[1], 2), "%")
#Task8b:FY2024 months with inflation above 25 %.
high_inflation = cpi_months[cpi[0] > 25]

print("Task8b:")
print("Months with inflation above 25%:")
print(high_inflation)
#Task8c: Drop in inflation for each month (FY2024 − FY2025).
inflation_drop = cpi[0] - cpi[1]

print("Task8c:")
print("Months  Inflation Drop")
for i in range(len(cpi_months)):
    print(cpi_months[i], ":", round(inflation_drop[i], 2), "%")
#Task8d:The month with the biggest drop.
biggest_index = np.argmax(inflation_drop)

print("Task8d:")
print("Month with the biggest drop:", cpi_months[biggest_index])
print("Biggest drop:", inflation_drop[biggest_index], "%")
#Task8e: Label each FY2025 month "Low" (< 5), "Moderate" (5–10) or "High" (≥ 10) with np.where.
labels = np.where(
    cpi[1] < 5,
    "Low",
    np.where(cpi[1] < 10, "Moderate", "High")
)
print("Task8e:")
print("Months Lables")
for i in range(len(cpi_months)):
    print(cpi_months[i], ":", labels[i])
```

##### Interpret: in 1–2 lines, what happened to prices in Pakistan between the two years?

Pakistan's average inflation decreased from 26.22% in FY2024 to 4.87% in FY2025, meaning prices increased much more slowly in FY2025.
However, prices did not become cheaper; the rate of price increases simply slowed down.

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# # D4. Marks of one lab section (sample class data)
marks = np.array([45, 78, 92, 38, 66, 97, 49, 55, 71, 83])
#Task9a:Add 5 grace marks but cap at 100 (np.clip).
grace_marks = np.clip(marks + 5, 0, 100)

print("Task9a:")
print("Marks after grace:", grace_marks)
#Task9b: Pass % (≥ 50) after grace.
passed = grace_marks >= 50

pass_percentage = np.mean(passed) * 100

print("Task9b:")
print("Pass Percentage:", pass_percentage, "%")
#Task9c:Index of students who passed only because of the grace marks.
grace_passed = np.where((marks < 50) & (grace_marks >= 50))[0]

print("Task9c:")
print("Students who passed only because of grace marks:", grace_passed)
#Task9d:z-score of each student after grace.
mean_marks = np.mean(grace_marks)
std_marks = np.std(grace_marks)

z_scores = (grace_marks - mean_marks) / std_marks

print("Task9d:")
print("Mean:", round(mean_marks, 2))
print("Standard Deviation:", round(std_marks, 2))
print("Z-Scores:", np.round(z_scores, 2))
```

```python
# Task10a:Start from np.zeros((8, 8), dtype=int) and draw the "7" from the lecture with slicing (white = 255, grey = 128).
import numpy as np

img = np.zeros((8, 8), dtype=int)

# Draw the top horizontal line of 7 (white)
img[1, 1:7] = 255

# Draw the diagonal line of 7 (grey)
img[2, 5:7] = 128
img[3, 4:6] = 128
img[4, 3:5] = 128
img[5, 2:4] = 128
img[6, 1:3] = 128

print("Task10a:")
print("Original Image:")
print(img)
#Task10b:Flip it left-right.
flipped_img = np.fliplr(img)

print("Task10b:")
print("Flipped Image:")
print(flipped_img)
#Task10c: Invert it (255 - img).
inverted_img = 255 - flipped_img

print("Task10c:")
print("Inverted Image:")
print(inverted_img)
#Task10d: Crop rows 1–4 and columns 3–6.
cropped_img = inverted_img[1:5, 3:7]

print("Task10d:")
print("Cropped Image:")
print(cropped_img)
#Task10e:Brighten by 80, capped at 255.
bright_img = np.clip(cropped_img + 80, 0, 255)

print("Task10e:")
print("Brightened Image:")
print(bright_img)
#Task10f:Print its shape, dtype and number of white pixels.
print("Task10f:")
print("Shape:", bright_img.shape)
print("Data Type:", bright_img.dtype)
print("Number of White Pixels:", np.sum(bright_img == 255))
```
